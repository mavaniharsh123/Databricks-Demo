```python
from pyspark.sql.functions import *
# Sample DataFrame with an array column
data = [
    (1, "Sagar", ["maths", "science", "history"],[20, 30, 40]),
    (2, "Alex", ["maths", "science",  "history"],[40, 20,45]),
    (3, "John", ["maths, science", "history"], [30,20,10]) 
]
columns = ["id", "name","subjects", "scores"]
df = spark.createDataFrame(data, columns)

# Explode the "scores" column
df_exploded = df.select("id", "name", explode("subjects").alias("subject"), explode("scores").alias("score")) #to over this subject and score we can use array_zip
df_exploded.show()


df.withColumn("new_mark", arrays_zip("subjects","scores")).display()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, monotonically_increasing_id

def main():
    spark = SparkSession.builder.appName("SortSales").getOrCreate()

    data = [
        ("A", 100),
        ("B", 200),
        ("C", 100),
        ("D", 300)
    ]

    # ✅ FIXED HERE
    df = spark.createDataFrame(data, ["product", "sales_amount"])

    df_with_index = df.withColumn("idx", monotonically_increasing_id())

    result_df = df_with_index.orderBy(
        col("sales_amount").desc(),
        col("idx").asc()
    ).drop("idx")

    result_df.show()

main()
```
